#### Approach
نستخدم `np.random.randint` لإنشاء مصفوفة عشوائية 4×4 من نوع `uint8` مباشرة، ثم نصل للبكسل المطلوب بالفهرسة العادية، ونعدّل بكسلاً آخر بنفس الطريقة.

#### Reference Implementation
```python
import numpy as np

# 1. Create a 4x4 grayscale image with random values between 0 and 255
grayscale_image = np.random.randint(0, 256, (4, 4), dtype=np.uint8)
print("Grayscale image:\n", grayscale_image)

# 2. Print the shape
print(f"Shape: {grayscale_image.shape}")

# 3. Access the pixel at [1, 2]
pixel_value = grayscale_image[1, 2]
print(f"Pixel at (1, 2): {pixel_value}")

# 4. Modify the pixel at [0, 0] to white (255)
grayscale_image[0, 0] = 255

# 5. Print the final array
print("Final image:\n", grayscale_image)
```

#### Explanation
- `np.random.randint(0, 256, (4, 4), dtype=np.uint8)` يُنشئ مصفوفة 4×4 من أعداد صحيحة عشوائية بين 0 (متضمّن) و256 (غير متضمّن)، أي فعلياً بين 0 و255 — بالضبط نطاق `uint8`.
- الفهرسة `grayscale_image[1, 2]` تصل مباشرة إلى الصف رقم 1 والعمود رقم 2.
- إسناد قيمة جديدة لموقع محدد `grayscale_image[0, 0] = 255` يُعدّل المصفوفة في مكانها (in-place).

#### Expected Result
طباعة مصفوفة 4×4، ثم شكلها `(4, 4)`، ثم قيمة بكسل واحدة، ثم المصفوفة النهائية مع ظهور 255 في الزاوية العلوية اليسرى.

#### Common Mistakes
- نسيان تحديد `dtype=np.uint8` فتصبح المصفوفة من نوع `int64` الافتراضي.
- الخلط بين الترتيب `[row, col]` و `[x, y]` (في NumPy الترتيب هو صف ثم عمود).
- استخدام `randint(0, 255, ...)` بدل `randint(0, 256, ...)`، مما يستبعد القيمة 255 من الاحتمالات.

### Solution

<details>
<summary>💡 Show Solution</summary>

#### Approach
نبدأ بصورة سوداء بالكامل بواسطة `np.zeros`، ثم نستخدم الفهرسة المباشرة لتلوين قناة كاملة (Broadcasting)، ثم نُعدّل بكسلاً واحداً بكل قنواته الثلاث.

#### Reference Implementation
```python
import numpy as np

# 1. Create a 5x5 black RGB image
rgb_image = np.zeros((5, 5, 3), dtype=np.uint8)

# 2. Make the whole image pure blue (channel index 2 = Blue)
rgb_image[:, :, 2] = 255

# 3. Change the pixel at [2, 2] to white
rgb_image[2, 2] = [255, 255, 255]

# 4. Print the Red channel only
red_channel = rgb_image[:, :, 0]
print("Red channel:\n", red_channel)

# 5. Print the shape
print(f"Shape: {rgb_image.shape}")
```

#### Explanation
- `np.zeros((5, 5, 3), dtype=np.uint8)` يُنشئ مصفوفة بثلاثة أبعاد: الارتفاع، العرض، وعدد القنوات (3 لصورة RGB).
- `rgb_image[:, :, 2] = 255` يستخدم Broadcasting لتعيين القيمة 255 لكل بكسلات القناة الثالثة (الزرقاء هنا) دفعة واحدة دون حلقة.
- `rgb_image[2, 2] = [255, 255, 255]` يُعدّل البكسل الواحد بكل قنواته الثلاث معاً.

#### Expected Result
شكل `(5, 5, 3)`، القناة الحمراء المطبوعة كلها أصفار عدا الموقع `[2, 2]` الذي يظهر فيه 255.

#### Common Mistakes
- الخلط بين ترتيب القنوات RGB (المستخدَم هنا يدوياً) وBGR (المستخدَم في OpenCV — سنراه في التمرين القادم).
- محاولة تلوين قناة كاملة باستخدام حلقة `for` بدلاً من الفهرسة المباشرة الأبسط والأسرع.
- تمرير قيمة واحدة بدل قائمة من 3 قيم عند تعديل بكسل RGB كامل.

</details>
